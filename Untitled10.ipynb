```python
df = pd.read_csv("diabetic_data.csv", na_values=['?'],keep_default_na=False)

df["target"] = (df["readmitted"] == "<30").astype(int)

df = df[~df["discharge_disposition_id"].isin([11, 13, 14, 19, 20, 21])]
df = df[df["gender"] != "Unknown/Invalid"]

grupos = df["patient_nbr"]
# Adicionamos 'diag_1', 'diag_2', 'diag_3' à lista de remoção por serem códigos de texto do CID-9
df = df.drop(columns=["encounter_id", "patient_nbr", "weight", "readmitted",
                      "examide", "citoglipton",'payer_code', 'medical_specialty',
                      "diag_1", "diag_2", "diag_3"])

med_cols = ["metformin","repaglinide","nateglinide","chlorpropamide","glimepiride",
            "acetohexamide","glipizide","glyburide","tolbutamide","pioglitazone",
            "rosiglitazone","acarbose","miglitol","troglitazone","tolazamide","insulin",
            "glyburide-metformin","glipizide-metformin","glimepiride-pioglitazone",
            "metformin-rosiglitazone","metformin-pioglitazone"]
df[med_cols] = df[med_cols].replace({"No": 0, "Steady": 1, "Up": 2, "Down": 3})

df["max_glu_serum"] = df["max_glu_serum"].map({"None":0, "Norm":1, ">200":2, ">300":3})
df["A1Cresult"]     = df["A1Cresult"].map({"None":0, "Norm":1, ">7":2, ">8":3})

df["change"] = (df["change"] == "Ch").astype(int)
df["diabetesMed"] = (df["diabetesMed"] == "Yes").astype(int)
df["gender"] = (df["gender"] == "Male").astype(int)
df["age"] = df["age"].str.extract(r"\[(\d+)-")[0].astype(int) + 5

df["race"] = df["race"].fillna("Unknown")
df["race"] = df["race"].replace({"Asian": "Other", "Hispanic": "Other"})

race_dummies = pd.get_dummies(df["race"], prefix="race")
df = pd.concat([df.drop(columns="race"), race_dummies], axis=1)

from sklearn.model_selection import GroupShuffleSplit
x, y = df.drop(columns="target"), df["target"]

# Criando X maiúsculo para manter compatibilidade lógica caso seja usado
X = x

gss = GroupShuffleSplit(n_splits=1, test_size=0.3, random_state=42)
tr, tmp = next(gss.split(X, y, grupos.loc[X.index]))
print(X.shape, y.mean())

df
```

```python
from sklearn.model_selection import train_test_split as tts
x_train, x_test, y_train, y_test = tts(x, y, test_size=0.2, random_state=100)

print("Novo formato de x_train:", x_train.shape)
x_train
```

```python
from sklearn.model_selection import train_test_split as tts
from sklearn.ensemble import RandomForestClassifier as rfc
from xgboost import XGBClassifier as xgbc
from sklearn.metrics import (roc_auc_score as ras, average_precision_score as aps,
                             confusion_matrix as cm, classification_report as cs,
                             precision_recall_curve as prc)

# Ajustar nomes de colunas com caracteres especiais para o XGBoost
for d in (x_train, x_test):
    d.columns = d.columns.str.replace(r"[\[\]<>]", "_", regex=True)

x_tr, x_val, y_tr, y_val = tts(x_train, y_train, test_size=0.2, random_state=100)

# 1. Instanciar e treinar o RandomForest
rf_model = rfc(n_estimators=300, max_depth=12, min_samples_leaf=20,
               class_weight="balanced_subsample", n_jobs=-1, random_state=42)
rf_model.fit(x_tr, y_tr)

# 2. Instanciar e treinar o XGBoost
peso = (y_tr == 0).sum() / (y_tr == 1).sum()
xgb_model = xgbc(n_estimators=400, max_depth=4, learning_rate=0.05,
                 subsample=0.8, colsample_bytree=0.8,
                 scale_pos_weight=peso, eval_metric="auc", random_state=42)
xgb_model.fit(x_tr, y_tr)

# 3. Avaliar os modelos instanciados e treinados
for nome, m in [("RandomForest", rf_model), ("XGBoost", xgb_model)]:
    p = m.predict_proba(x_val)[:, 1]
    print(f"{nome}: AUC-ROC={ras(y_val, p):.3f}  AUC-PR={aps(y_val, p):.3f}")
```

```python
from sklearn.model_selection import RandomizedSearchCV, GroupKFold
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from scipy.stats import randint, uniform

# Garantindo que os dados de grupos estejam alinhados
g_tr = grupos.loc[x_tr.index]
gkf = GroupKFold(n_splits=3)

# Redefinindo os estimadores base
rf_base = RandomForestClassifier(class_weight="balanced_subsample", n_jobs=-1, random_state=42)
rf_space = {
    "n_estimators": randint(150, 350),
    "max_depth": randint(6, 12),
    "min_samples_leaf": randint(10, 40),
    "max_features": ["sqrt", 0.3]
}

xgb_base = XGBClassifier(scale_pos_weight=peso, eval_metric="auc", random_state=42, n_jobs=-1)
xgb_space = {
    "n_estimators": randint(150, 350),
    "max_depth": randint(3, 6),
    "learning_rate": uniform(0.02, 0.08),
    "subsample": uniform(0.7, 0.3),
    "colsample_bytree": uniform(0.7, 0.3)
}

# Usando n_jobs=-1 e n_iter=5 para garantir execução rápida e eficiente
buscas = {}
for nome, base, space in [("RF", rf_base, rf_space), ("XGB", xgb_base, xgb_space)]:
    print(f"Iniciando sintonia fina para {nome}...")
    b = RandomizedSearchCV(base, space, n_iter=5, scoring="average_precision",
                           cv=gkf, n_jobs=-1, random_state=42, verbose=1)
    b.fit(x_tr, y_tr, groups=g_tr)
    buscas[nome] = b
    print(nome, "melhor AUC-PR (CV):", round(b.best_score_, 3))
    print("Melhores parâmetros:", b.best_params_)

rf_final, xgb_final = buscas["RF"].best_estimator_, buscas["XGB"].best_estimator_
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay, roc_curve, precision_recall_curve, classification_report, roc_auc_score, average_precision_score
import seaborn as sns
import pandas as pd
import numpy as np

# Ajustar nomes de colunas do conjunto de teste para evitar incompatibilidade de caracteres especiais no XGBoost
x_test.columns = x_test.columns.str.replace(r'[\[\]<>]', '_', regex=True)

# Dicionário dos modelos otimizados finais
modelos = {
    "Random Forest (Otimizado)": rf_final,
    "XGBoost (Otimizado)": xgb_final
}

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, (nome, modelo) in zip(axes, modelos.items()):
    p_test = modelo.predict_proba(x_test)[:, 1]
    y_pred = (p_test >= 0.5).astype(int)

    # Matriz de Confusão
    cm_val = cm(y_test, y_pred)
    ConfusionMatrixDisplay(confusion_matrix=cm_val, display_labels=["Não Readmitido", "Readmitido"]).plot(ax=ax, cmap="Blues", values_format="d")
    ax.set_title(f"Matriz de Confusão - {nome}")



plt.tight_layout()
plt.show()

# Exibindo métricas completas e AUCs no Teste
for nome, modelo in modelos.items():
    p_test = modelo.predict_proba(x_test)[:, 1]
    auc_roc = roc_auc_score(y_test, p_test)
    auc_pr = average_precision_score(y_test, p_test)
    print(f"=== {nome} no Teste Final ===")
    print(f"AUC-ROC: {auc_roc:.3f} | AUC-PR: {auc_pr:.3f}\n")
```

```python
# Analisando o impacto clínico de Falsos Positivos e Falsos Negativos
for nome, modelo in modelos.items():
    p_test = modelo.predict_proba(x_test)[:, 1]
    y_pred = (p_test >= 0.5).astype(int)
    cm_val = cm(y_test, y_pred)

    tn, fp, fn, tp = cm_val.ravel()
    print(f"=== Análise de Erros: {nome} ===")
    print(f"Falsos Positivos (FPs): {fp} pacientes foram erroneamente previstos como retorno iminente.")

    print(f"Falsos Negativos (FNs): {fn} pacientes que realmente retornariam não foram detectados pelo modelo.")
```

```python
# Comparação quanto ao Balanço de Classes: Com vs. Sem peso
rf_sem = RandomForestClassifier(n_estimators=200, max_depth=10, min_samples_leaf=17, max_features=0.3, random_state=42, n_jobs=-1)
xgb_sem = XGBClassifier(n_estimators=237, max_depth=5, learning_rate=0.03, colsample_bytree=0.74, subsample=0.8, scale_pos_weight=1, eval_metric="auc", random_state=42, n_jobs=-1)

print("Treinando modelos sem balanceamento de classes...")
rf_sem.fit(x_tr, y_tr)
xgb_sem.fit(x_tr, y_tr)

print("\n--- COMPARAÇÃO DE MODELOS COM VS SEM BALANCEAMENTO (Métricas no Teste) ---")
for rotulo, mod_com, mod_sem in [("Random Forest", rf_final, rf_sem), ("XGBoost", xgb_final, xgb_sem)]:
    p_com = mod_com.predict_proba(x_test)[:, 1]
    p_sem = mod_sem.predict_proba(x_test)[:, 1]

    y_pred_com = (p_com >= 0.5).astype(int)
    y_pred_sem = (p_sem >= 0.5).astype(int)

    print(f"\n[{rotulo} Otimizado - COM Balanceamento]")
    print(f"  Revocação (Sensibilidade): {cs(y_test, y_pred_com, output_dict=True)['1']['recall']:.3f}")
    print(f"  AUC-PR: {average_precision_score(y_test, p_com):.3f}")

    print(f"[{rotulo} Otimizado - SEM Balanceamento]")
    print(f"  Revocação (Sensibilidade): {cs(y_test, y_pred_sem, output_dict=True)['1']['recall']:.3f}")
    print(f"  AUC-PR: {average_precision_score(y_test, p_sem):.3f}")
```
